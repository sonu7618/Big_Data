```python
import os
from pyspark.sql import SparkSession
from pyspark.sql.functions import col

os.makedirs("/home/sonu/BIG_DATA/outputs", exist_ok=True)
os.makedirs("/home/sonu/BIG_DATA/data/processed", exist_ok=True)

spark = SparkSession.builder \
    .appName("BusComplianceCleaning") \
    .config("spark.hadoop.fs.defaultFS", "file:///") \
    .config("spark.sql.shuffle.partitions", "8") \
    .getOrCreate()
```

```python
timetables_df = spark.read.csv(
    "file:///home/sonu/BIG_DATA/data/timetables_csv/timetables_combined.csv",
    header=True, inferSchema=True
)
location_df = spark.read.csv(
    "file:///home/sonu/BIG_DATA/data/location_data/location_history.csv",
    header=True, inferSchema=True
)

print("timetables:", timetables_df.count())
print("location:", location_df.count())
```

```python
location_df.select("recorded_at_time").show(5, truncate=False)
location_df.printSchema()
```

```python
timetables_clean = timetables_df.dropna(subset=["stop_ref", "latitude", "longitude", "scheduled_time"])

timetables_clean = timetables_clean.withColumn("latitude", col("latitude").cast("double")) \
                                    .withColumn("longitude", col("longitude").cast("double"))

print("timetables after cleaning:", timetables_clean.count())
```

```python
location_clean = location_df.dropna(subset=["latitude", "longitude", "line_ref", "recorded_at_time"])

location_clean = location_clean.withColumn("latitude", col("latitude").cast("double")) \
                                .withColumn("longitude", col("longitude").cast("double"))

# recorded_at_time already came through as a proper timestamp type (confirmed in Cell 3),
# so just reuse it directly instead of trying to re-parse an already-parsed value
location_clean = location_clean.withColumn("recorded_at_ts", col("recorded_at_time"))

location_clean.select("recorded_at_time", "recorded_at_ts").show(10, truncate=False)

null_check = location_clean.filter(col("recorded_at_ts").isNull()).count()
print(f"rows where recorded_at_ts failed to parse: {null_check} out of {location_clean.count()}")
```

```python
location_clean = location_clean.dropDuplicates(["vehicle_ref", "recorded_at_time", "line_ref"])
print("location after dedupe:", location_clean.count())
```

```python
timetables_clean.show(5, truncate=False)
location_clean.show(5, truncate=False)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_style("whitegrid")
```

```python
labels = ["Timetables\nRaw", "Timetables\nCleaned", "Location\nRaw", "Location\nCleaned"]
values = [timetables_df.count(), timetables_clean.count(), location_df.count(), location_clean.count()]

fig, ax = plt.subplots(figsize=(9, 5))
ax.bar(labels, values, color=["#4C72B0", "#55A868", "#4C72B0", "#55A868"])
ax.set_title("Figure 1: Row Counts Before and After Cleaning")
ax.set_xlabel("Dataset Stage")
ax.set_ylabel("Row Count")
plt.tight_layout()
plt.savefig("/home/sonu/BIG_DATA/outputs/fig1_cleaning_before_after.png")
plt.show()
```

```python
timetables_clean.write.mode("overwrite").parquet("file:///home/sonu/BIG_DATA/data/processed/timetables_clean.parquet")
location_clean.write.mode("overwrite").parquet("file:///home/sonu/BIG_DATA/data/processed/location_clean.parquet")
print("saved")
```
