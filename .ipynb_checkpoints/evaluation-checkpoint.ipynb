```python
import os
from pyspark.sql import SparkSession
from pyspark.ml.evaluation import BinaryClassificationEvaluator, MulticlassClassificationEvaluator
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

os.makedirs("/home/sonu/BIG_DATA/outputs", exist_ok=True)

spark = SparkSession.builder \
    .appName("BusComplianceEvaluation") \
    .config("spark.hadoop.fs.defaultFS", "file:///") \
    .getOrCreate()
```

```python
lr_predictions = spark.read.parquet("file:///home/sonu/BIG_DATA/data/processed/lr_predictions.parquet")
rf_predictions = spark.read.parquet("file:///home/sonu/BIG_DATA/data/processed/rf_predictions.parquet")
gbt_predictions = spark.read.parquet("file:///home/sonu/BIG_DATA/data/processed/gbt_predictions.parquet")
```

```python
auc_eval = BinaryClassificationEvaluator(labelCol="label", metricName="areaUnderROC")
acc_eval = MulticlassClassificationEvaluator(labelCol="label", metricName="accuracy")
prec_eval = MulticlassClassificationEvaluator(labelCol="label", metricName="weightedPrecision")
rec_eval = MulticlassClassificationEvaluator(labelCol="label", metricName="weightedRecall")
f1_eval = MulticlassClassificationEvaluator(labelCol="label", metricName="f1")

def evaluate(predictions, name):
    return {
        "model": name,
        "accuracy": acc_eval.evaluate(predictions),
        "precision": prec_eval.evaluate(predictions),
        "recall": rec_eval.evaluate(predictions),
        "f1": f1_eval.evaluate(predictions),
        "roc_auc": auc_eval.evaluate(predictions),
    }
```

```python
results = [
    evaluate(lr_predictions, "Logistic Regression"),
    evaluate(rf_predictions, "Random Forest"),
    evaluate(gbt_predictions, "Gradient-Boosted Trees"),
]
results_df = pd.DataFrame(results)
results_df
```

```python
fig, ax = plt.subplots(figsize=(10, 6))
metrics = ["accuracy", "precision", "recall", "f1", "roc_auc"]
x = np.arange(len(metrics))
width = 0.25

for i, row in results_df.iterrows():
    ax.bar(x + i * width, [row[m] for m in metrics], width, label=row["model"])

ax.set_title("Figure 1: Model Performance Comparison")
ax.set_xlabel("Metric")
ax.set_ylabel("Score")
ax.set_xticks(x + width)
ax.set_xticklabels(metrics)
ax.legend(title="Model")
plt.tight_layout()
plt.savefig("/home/sonu/BIG_DATA/outputs/fig1_model_comparison.png")
plt.show()
```

```python
def plot_confusion_matrix(ax, predictions_df, title):
    counts = predictions_df.groupBy("label", "prediction").count().toPandas()
    matrix = np.zeros((2, 2))
    for _, row in counts.iterrows():
        matrix[int(row["label"])][int(row["prediction"])] = row["count"]

    labels = ["Not Compliant", "Compliant"]
    ax.imshow(matrix, cmap="Blues")
    for i in range(2):
        for j in range(2):
            value = int(matrix[i, j])
            color = "white" if value > matrix.max() / 2 else "black"
            ax.text(j, i, f"{value:,}", ha="center", va="center", color=color, fontsize=11)
    ax.set_xticks([0, 1]); ax.set_yticks([0, 1])
    ax.set_xticklabels(labels); ax.set_yticklabels(labels)
    ax.set_xlabel("Predicted"); ax.set_ylabel("Actual")
    ax.set_title(title)

fig, axes = plt.subplots(1, 3, figsize=(16, 5))
plot_confusion_matrix(axes[0], lr_predictions, "Logistic Regression")
plot_confusion_matrix(axes[1], rf_predictions, "Random Forest")
plot_confusion_matrix(axes[2], gbt_predictions, "Gradient-Boosted Trees")
plt.suptitle("Figure 2: Confusion Matrices Across All Models")
plt.tight_layout()
plt.savefig("/home/sonu/BIG_DATA/outputs/fig2_confusion_matrices.png")
plt.show()
```

```python
results_df.to_csv("/home/sonu/BIG_DATA/outputs/model_comparison.csv", index=False)
print("saved")
```
