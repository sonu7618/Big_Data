```python
import os
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, hour, dayofweek, avg
from pyspark.ml.feature import StringIndexer, VectorAssembler

os.makedirs("/home/sonu/BIG_DATA/outputs", exist_ok=True)

spark = SparkSession.builder \
    .appName("BusComplianceFeatures") \
    .config("spark.hadoop.fs.defaultFS", "file:///") \
    .getOrCreate()
```

```python
df = spark.read.parquet("file:///home/sonu/BIG_DATA/data/processed/joined_compliance.parquet")
print("rows:", df.count())
df.printSchema()
```

```python
row_count = df.count()
if row_count == 0:
    raise ValueError("STOP: joined_compliance.parquet is empty. Fix notebook 02/03 first.")

null_check = df.filter(col("scheduled_datetime").isNull()).count()
print(f"rows: {row_count}, null scheduled_datetime: {null_check}")

if null_check == row_count:
    raise ValueError("STOP: every row has null scheduled_datetime. Fix notebook 02/03 first.")

df = df.dropna(subset=["scheduled_datetime", "deviation_minutes"])
print("rows after dropping remaining nulls:", df.count())
```

```python
df = df.withColumn("hour_of_day", hour(col("scheduled_datetime")))
df = df.withColumn("day_of_week", dayofweek(col("scheduled_datetime")))
df.select("scheduled_datetime", "hour_of_day", "day_of_week").show(5)
```

```python
route_avg_delay = df.groupBy("line_name").agg(avg("deviation_minutes").alias("route_avg_delay"))

if "route_avg_delay" in df.columns:
    df = df.drop("route_avg_delay")

df = df.join(route_avg_delay, on="line_name", how="left")
df.select("line_name", "route_avg_delay").distinct().show(10)
```

```python
line_indexer = StringIndexer(inputCol="line_name", outputCol="line_index", handleInvalid="keep")
df = line_indexer.fit(df).transform(df)
```

```python
feature_cols = ["hour_of_day", "day_of_week", "route_avg_delay", "line_index"]

assembler = VectorAssembler(inputCols=feature_cols, outputCol="features", handleInvalid="skip")
df_features = assembler.transform(df)

df_features.select("features", "compliant").show(5, truncate=False)
print("final feature row count:", df_features.count())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_style("whitegrid")
```

```python
hourly = df.groupBy("hour_of_day").avg("deviation_minutes").orderBy("hour_of_day").dropna().toPandas()

fig, ax = plt.subplots(figsize=(9, 5))
ax.plot(hourly["hour_of_day"], hourly["avg(deviation_minutes)"], marker="o", color="#8172B2")
ax.set_title("Figure 1: Average Delay by Hour of Day")
ax.set_xlabel("Hour of Day (24h)")
ax.set_ylabel("Average Deviation (minutes)")
plt.tight_layout()
plt.savefig("/home/sonu/BIG_DATA/outputs/fig1_hourly_delay_line.png")
plt.show()
```

```python
route_delay_pd = route_avg_delay.orderBy("route_avg_delay", ascending=False).dropna().limit(15).toPandas()

fig, ax = plt.subplots(figsize=(10, 6))
ax.bar(route_delay_pd["line_name"], route_delay_pd["route_avg_delay"], color="#C44E52")
ax.set_title("Figure 2: Average Delay by Route (Top 15)")
ax.set_xlabel("Route")
ax.set_ylabel("Average Deviation (minutes)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("/home/sonu/BIG_DATA/outputs/fig2_route_delay_bar.png")
plt.show()
```

```python
df_features.write.mode("overwrite").parquet("file:///home/sonu/BIG_DATA/data/processed/features.parquet")
print("saved")
```
