```python
from pyspark.sql import SparkSession
from pyspark.ml.classification import RandomForestClassificationModel

spark = SparkSession.builder \
    .appName("BusCompliancePredictExample") \
    .config("spark.hadoop.fs.defaultFS", "file:///") \
    .getOrCreate()
```

```python
model = RandomForestClassificationModel.load("file:///home/sonu/BIG_DATA/data/processed/models/rf_model")
features_df = spark.read.parquet("file:///home/sonu/BIG_DATA/data/processed/features.parquet")
```

```python
example = features_df.limit(1)
example.select("line_name", "stop_name", "hour_of_day", "day_of_week", "route_avg_delay", "compliant").show(truncate=False)
```

```python
prediction = model.transform(example)
prediction.select("line_name", "stop_name", "compliant", "prediction", "probability").show(truncate=False)
```

```python
import matplotlib.pyplot as plt

prob_row = prediction.select("probability").collect()[0][0]
labels = ["Not Compliant", "Compliant"]

fig, ax = plt.subplots(figsize=(6, 5))
ax.bar(labels, prob_row, color=["#C44E52", "#55A868"])
ax.set_title("Figure 1: Predicted Probability for This Trip")
ax.set_xlabel("Class")
ax.set_ylabel("Probability")
plt.tight_layout()
plt.savefig("/home/sonu/BIG_DATA/outputs/fig1_predict_example_probability.png")
plt.show()
```
