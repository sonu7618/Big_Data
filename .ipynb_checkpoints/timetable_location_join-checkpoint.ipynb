```python
import os
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, to_timestamp, unix_timestamp, abs as spark_abs, when, concat_ws, upper, trim, regexp_replace, count, date_format

os.makedirs("/home/sonu/BIG_DATA/outputs", exist_ok=True)

spark = SparkSession.builder \
    .appName("BusComplianceJoin") \
    .config("spark.hadoop.fs.defaultFS", "file:///") \
    .config("spark.sql.shuffle.partitions", "8") \
    .getOrCreate()
```

```python
timetables = spark.read.parquet("file:///home/sonu/BIG_DATA/data/processed/timetables_clean.parquet")
location = spark.read.parquet("file:///home/sonu/BIG_DATA/data/processed/location_clean.parquet")

print("timetables:", timetables.count())
print("location:", location.count())
```

```python
location = location.withColumn("recorded_date", col("recorded_at_ts").cast("date"))
location.select("recorded_at_ts", "recorded_date").show(5, truncate=False)
```

```python
timetables = timetables.repartition(4, "line_name")
location = location.repartition(4, "line_ref")

timetables.cache()
location.cache()

timetables.count()
location.count()
```

```python
from pyspark.sql.functions import date_format

timetables_for_join = timetables.drop("latitude", "longitude") \
    .withColumn("line_name_clean", trim(upper(col("line_name")))) \
    .withColumn("scheduled_time_only", date_format(col("scheduled_time"), "HH:mm:ss"))

location_for_join = location.withColumn(
    "line_ref_clean",
    trim(upper(regexp_replace(col("line_ref"), r"^[A-Z]+[:\-_]", "")))
)

joined = timetables_for_join.join(
    location_for_join,
    timetables_for_join.line_name_clean == location_for_join.line_ref_clean,
    how="inner"
)

print("joined row count (before time filter):", joined.count())
joined.show(5, truncate=False)
```

```python
joined = joined.withColumn(
    "scheduled_datetime",
    to_timestamp(concat_ws(" ", col("recorded_date"), col("scheduled_time_only")))
)

# only keep matches where the GPS ping is within 60 minutes of the scheduled time -
# this is what cuts the row count from millions down to a realistic number
joined_ts = joined.filter(
    spark_abs(unix_timestamp(col("recorded_at_ts")) - unix_timestamp(col("scheduled_datetime"))) <= 3600
)

print("joined row count (after time filter):", joined_ts.count())
joined_ts.select("recorded_date", "scheduled_time_only", "scheduled_datetime", "recorded_at_ts").show(10, truncate=False)
```

```python
joined_ts = joined_ts.withColumn(
    "deviation_minutes",
    spark_abs(unix_timestamp(col("recorded_at_ts")) - unix_timestamp(col("scheduled_datetime"))) / 60
)

joined_ts.select("line_name", "stop_name", "scheduled_datetime", "recorded_at_ts", "deviation_minutes").show(10, truncate=False)
```

```python
joined_ts.select(
    count(when(col("deviation_minutes").isNull(), True)).alias("null_deviation"),
    count(when(col("scheduled_datetime").isNull(), True)).alias("null_scheduled_dt"),
    count("*").alias("total_rows")
).show()
```

```python
COMPLIANCE_THRESHOLD_MINUTES = 2

joined_labelled = joined_ts.withColumn(
    "compliant",
    when(col("deviation_minutes") <= COMPLIANCE_THRESHOLD_MINUTES, 1).otherwise(0)
)

joined_labelled.groupBy("compliant").count().show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_style("whitegrid")
```

```python
compliance_counts = joined_labelled.groupBy("compliant").count().toPandas()
compliance_counts["label"] = compliance_counts["compliant"].map({1: "Compliant", 0: "Non-Compliant"})

fig, ax = plt.subplots(figsize=(6, 6))
ax.pie(compliance_counts["count"], labels=compliance_counts["label"], autopct="%1.1f%%",
       colors=["#55A868", "#C44E52"], startangle=90)
ax.set_title("Figure 1: Compliance Split After Join")
plt.tight_layout()
plt.savefig("/home/sonu/BIG_DATA/outputs/fig1_join_compliance_pie.png")
plt.show()
```

```python
deviation_sample = joined_labelled.select("deviation_minutes").dropna().sample(0.1, seed=42).toPandas()

fig, ax = plt.subplots(figsize=(9, 5))
ax.hist(deviation_sample["deviation_minutes"], bins=40, color="#DD8452", edgecolor="white")
ax.set_title("Figure 2: Distribution of Schedule Deviation (Post-Join)")
ax.set_xlabel("Deviation from Schedule (minutes)")
ax.set_ylabel("Frequency")
plt.tight_layout()
plt.savefig("/home/sonu/BIG_DATA/outputs/fig2_join_deviation_histogram.png")
plt.show()
```

```python
joined_labelled.write.mode("overwrite").parquet("file:///home/sonu/BIG_DATA/data/processed/joined_compliance.parquet")
print("saved")
```
