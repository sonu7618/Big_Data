```python
import os
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, count, when

os.makedirs("/home/sonu/BIG_DATA/outputs", exist_ok=True)

spark = SparkSession.builder \
    .appName("BusComplianceExploration") \
    .config("spark.hadoop.fs.defaultFS", "file:///") \
    .config("spark.sql.shuffle.partitions", "8") \
    .getOrCreate()
spark
```

```python
timetables_df = spark.read.csv(
    "file:///home/sonu/BIG_DATA/data/timetables_csv/timetables_combined.csv",
    header=True, inferSchema=True
)
timetables_df.printSchema()
```

```python
timetable_row_count = timetables_df.count()
print(f"timetable rows: {timetable_row_count}")
timetables_df.show(5, truncate=False)
```

```python
location_df = spark.read.csv(
    "file:///home/sonu/BIG_DATA/data/location_data/location_history.csv",
    header=True, inferSchema=True
)
location_df.printSchema()
```

```python
location_row_count = location_df.count()
print(f"location rows: {location_row_count}")
location_df.show(5, truncate=False)
```

```python
total_records = timetable_row_count + location_row_count
print(f"timetable: {timetable_row_count}")
print(f"location: {location_row_count}")
print(f"combined total: {total_records}")
print(f"meets 100k threshold: {total_records >= 100000}")
```

```python
def null_counts(df, name):
    print(f"--- null counts: {name} ---")
    df.select([count(when(col(c).isNull(), c)).alias(c) for c in df.columns]).show(truncate=False)

null_counts(timetables_df, "timetables")
null_counts(location_df, "location")
```

```python
print("distinct lines in timetables:", timetables_df.select("line_name").distinct().count())
print("distinct stops in timetables:", timetables_df.select("stop_ref").distinct().count())
print("distinct lines in location data:", location_df.select("line_ref").distinct().count())
print("distinct vehicles in location data:", location_df.select("vehicle_ref").distinct().count())
```

```python
timetables_df.select("latitude", "longitude").describe().show()
location_df.select("latitude", "longitude", "bearing").describe().show()
```

```python
timetable_lines = set(row.line_name for row in timetables_df.select("line_name").distinct().collect())
location_lines = set(row.line_ref for row in location_df.select("line_ref").distinct().collect())

overlap = timetable_lines & location_lines
print(f"lines in timetables: {len(timetable_lines)}")
print(f"lines in location data: {len(location_lines)}")
print(f"lines in both: {len(overlap)}")
print(sorted(overlap)[:20])
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from pyspark.sql import functions as F
sns.set_style("whitegrid")
```

```python
route_counts = timetables_df.groupBy("line_name").count().orderBy("count", ascending=False).limit(15).toPandas()

fig, ax = plt.subplots(figsize=(10, 6))
ax.bar(route_counts["line_name"], route_counts["count"], color="#4C72B0")
ax.set_title("Figure 1: Scheduled Stop-Time Records by Route (Top 15)")
ax.set_xlabel("Route (line_name)")
ax.set_ylabel("Number of Scheduled Stop Records")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("/home/sonu/BIG_DATA/outputs/fig1_trips_by_route.png")
plt.show()
```

```python
stops_per_route = timetables_df.groupBy("line_name").agg(F.countDistinct("stop_ref").alias("distinct_stops")).toPandas()

fig, ax = plt.subplots(figsize=(9, 5))
ax.hist(stops_per_route["distinct_stops"], bins=20, color="#55A868", edgecolor="white")
ax.set_title("Figure 2: Distribution of Distinct Stops per Route")
ax.set_xlabel("Number of Distinct Stops")
ax.set_ylabel("Number of Routes")
plt.tight_layout()
plt.savefig("/home/sonu/BIG_DATA/outputs/fig2_stops_per_route_histogram.png")
plt.show()
```

```python
labels = ["Timetable Records", "Location (GPS) Records"]
sizes = [timetable_row_count, location_row_count]

fig, ax = plt.subplots(figsize=(6, 6))
ax.pie(sizes, labels=labels, autopct="%1.1f%%", colors=["#4C72B0", "#C44E52"], startangle=90)
ax.set_title("Figure 3: Proportion of Timetable vs Location Records")
plt.tight_layout()
plt.savefig("/home/sonu/BIG_DATA/outputs/fig3_record_split_pie.png")
plt.show()
```

```python
stops_geo = timetables_df.select("latitude", "longitude").dropna().sample(0.2, seed=42).toPandas()

fig, ax = plt.subplots(figsize=(7, 7))
ax.scatter(stops_geo["longitude"], stops_geo["latitude"], s=6, alpha=0.4, color="#8172B2")
ax.set_title("Figure 4: Geographic Spread of Scheduled Stops")
ax.set_xlabel("Longitude")
ax.set_ylabel("Latitude")
plt.tight_layout()
plt.savefig("/home/sonu/BIG_DATA/outputs/fig4_stops_scatter.png")
plt.show()
```

```python
categories = ["Timetable\nLines", "Timetable\nStops", "Location\nLines", "Location\nVehicles"]
values = [
    timetables_df.select("line_name").distinct().count(),
    timetables_df.select("stop_ref").distinct().count(),
    location_df.select("line_ref").distinct().count(),
    location_df.select("vehicle_ref").distinct().count(),
]

fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(categories, values, marker="o", color="#DD8452", linewidth=2)
ax.set_title("Figure 5: Cardinality Summary Across Datasets")
ax.set_xlabel("Category")
ax.set_ylabel("Distinct Count")
plt.tight_layout()
plt.savefig("/home/sonu/BIG_DATA/outputs/fig5_cardinality_summary.png")
plt.show()
```
