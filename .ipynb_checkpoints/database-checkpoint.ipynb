```python
from pyspark.sql import SparkSession
import sqlite3
import os

spark = SparkSession.builder \
    .appName("BusComplianceDatabase") \
    .config("spark.hadoop.fs.defaultFS", "file:///") \
    .getOrCreate()
```

```python
df = spark.read.parquet("file:///home/sonu/BIG_DATA/data/processed/joined_compliance.parquet")
pdf = df.select(
    "line_name", "stop_name", "latitude", "longitude",
    "scheduled_datetime", "recorded_at_ts", "deviation_minutes", "compliant"
).toPandas()

pdf.head()
```

```python
os.makedirs("/home/sonu/BIG_DATA/db", exist_ok=True)
conn = sqlite3.connect("/home/sonu/BIG_DATA/db/bus_compliance.db")
cur = conn.cursor()

cur.execute("""
CREATE TABLE IF NOT EXISTS compliance_records (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    line_name TEXT,
    stop_name TEXT,
    latitude REAL,
    longitude REAL,
    scheduled_datetime TEXT,
    recorded_at TEXT,
    deviation_minutes REAL,
    compliant INTEGER
)
""")
conn.commit()
```

```python
insert_query = """
INSERT INTO compliance_records
(line_name, stop_name, latitude, longitude, scheduled_datetime, recorded_at, deviation_minutes, compliant)
VALUES (?, ?, ?, ?, ?, ?, ?, ?)
"""

rows = [
    (
        r.line_name, r.stop_name, r.latitude, r.longitude,
        str(r.scheduled_datetime), str(r.recorded_at_ts),
        r.deviation_minutes, int(r.compliant)
    )
    for r in pdf.itertuples()
]

cur.executemany(insert_query, rows)
conn.commit()

print(f"inserted {len(rows)} rows")
```

```python
sample_query = """
SELECT line_name, AVG(compliant) as compliance_rate, COUNT(*) as trip_count
FROM compliance_records
GROUP BY line_name
ORDER BY compliance_rate ASC
LIMIT 10
"""

cur.execute(sample_query)
for row in cur.fetchall():
    print(row)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_style("whitegrid")

cur.execute("SELECT line_name, AVG(compliant) as compliance_rate FROM compliance_records GROUP BY line_name ORDER BY compliance_rate ASC LIMIT 15")
rows = cur.fetchall()
lines = [r[0] for r in rows]
rates = [r[1] for r in rows]

fig, ax = plt.subplots(figsize=(10, 6))
ax.barh(lines, rates, color="#4C72B0")
ax.set_title("Figure 1: Compliance Rate by Route (from Database, Lowest 15)")
ax.set_xlabel("Compliance Rate")
ax.set_ylabel("Route")
plt.tight_layout()
plt.savefig("/home/sonu/BIG_DATA/outputs/fig1_db_compliance_by_route.png")
plt.show()
```

```python
with open("/home/sonu/BIG_DATA/docs/database_schema_dump.sql", "w") as f:
    for line in conn.iterdump():
        f.write(f"{line}\n")

print("schema dump saved to docs/database_schema_dump.sql")

conn.close()
```
