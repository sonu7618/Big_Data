```python
import os
from pyspark.sql import SparkSession
from pyspark.sql.functions import when, col
from pyspark.ml.classification import LogisticRegression, RandomForestClassifier, GBTClassifier

os.makedirs("/home/sonu/BIG_DATA/outputs", exist_ok=True)
os.makedirs("/home/sonu/BIG_DATA/data/processed/models", exist_ok=True)
spark.sparkContext.setLogLevel("ERROR")

spark = SparkSession.builder \
    .appName("BusComplianceModeling") \
    .config("spark.hadoop.fs.defaultFS", "file:///") \
    .config("spark.driver.memory", "4g") \
    .getOrCreate()
```

```python
df = spark.read.parquet("file:///home/sonu/BIG_DATA/data/processed/features.parquet")
df = df.withColumnRenamed("compliant", "label")
print("rows:", df.count())
df.printSchema()
```

```python
label_check = df.groupBy("label").count().collect()
label_dict = {row["label"]: row["count"] for row in label_check}
print("label distribution:", label_dict)

if 0 not in label_dict or 1 not in label_dict:
    raise ValueError("STOP: only one class present. Re-run notebook 04 first with the fixed 03 output.")
print("both classes present, safe to continue")
```

```python
train_df, test_df = df.randomSplit([0.8, 0.2], seed=42)
print("train:", train_df.count())
print("test:", test_df.count())
```

```python
train_labels = train_df.groupBy("label").count().collect()
train_dict = {row["label"]: row["count"] for row in train_labels}
print("train label distribution:", train_dict)

if 0 not in train_dict or 1 not in train_dict:
    raise ValueError("STOP: train split has only one class.")
print("both classes present in train_df")
```

```python
if "class_weight" in train_df.columns:
    train_df = train_df.drop("class_weight")

balance_ratio = train_dict[0] / train_dict[1]
print(f"balance ratio: {balance_ratio:.2f}")

train_df = train_df.withColumn("class_weight", when(col("label") == 1, balance_ratio).otherwise(1.0))
```

```python
lr = LogisticRegression(featuresCol="features", labelCol="label", weightCol="class_weight", maxIter=50)
lr_model = lr.fit(train_df)
lr_predictions = lr_model.transform(test_df)
lr_predictions.select("label", "prediction", "probability").show(5)
```

```python
rf = RandomForestClassifier(
    featuresCol="features",
    labelCol="label",
    numTrees=50,
    maxBins=64,   # must be >= the number of distinct routes (line_index categories) - 48 in your case, 64 gives headroom
    seed=42
)
rf_model = rf.fit(train_df)
rf_predictions = rf_model.transform(test_df)
rf_predictions.select("label", "prediction", "probability").show(5)
```

```python
gbt = GBTClassifier(
    featuresCol="features",
    labelCol="label",
    maxIter=20,   # reduced from 50 - still gives a meaningful comparison, trains much faster
    maxBins=64,
    seed=42
)
gbt_model = gbt.fit(train_df)
gbt_predictions = gbt_model.transform(test_df)
gbt_predictions.select("label", "prediction", "probability").show(5)
```

```python
print("Logistic Regression predictions:")
lr_predictions.groupBy("prediction").count().show()
print("Random Forest predictions:")
rf_predictions.groupBy("prediction").count().show()
print("GBT predictions:")
gbt_predictions.groupBy("prediction").count().show()
```

```python
lr_predictions.write.mode("overwrite").parquet("file:///home/sonu/BIG_DATA/data/processed/lr_predictions.parquet")
rf_predictions.write.mode("overwrite").parquet("file:///home/sonu/BIG_DATA/data/processed/rf_predictions.parquet")
gbt_predictions.write.mode("overwrite").parquet("file:///home/sonu/BIG_DATA/data/processed/gbt_predictions.parquet")

lr_model.write().overwrite().save("file:///home/sonu/BIG_DATA/data/processed/models/lr_model")
rf_model.write().overwrite().save("file:///home/sonu/BIG_DATA/data/processed/models/rf_model")
gbt_model.write().overwrite().save("file:///home/sonu/BIG_DATA/data/processed/models/gbt_model")

print("saved all three models and their predictions")
```
