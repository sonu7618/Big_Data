```python
from pyspark.sql import SparkSession
import matplotlib.pyplot as plt
import seaborn as sns

spark = SparkSession.builder \
    .appName("BusComplianceEDA") \
    .config("spark.hadoop.fs.defaultFS", "file:///") \
    .getOrCreate()
sns.set_style("whitegrid")
```

```python
df = spark.read.parquet("file:///home/sonu/BIG_DATA/data/processed/joined_compliance.parquet")
```

```python
route_compliance = df.groupBy("line_name").avg("compliant").toPandas()
route_compliance = route_compliance.sort_values("avg(compliant)")

plt.figure(figsize=(10, 6))
plt.barh(route_compliance["line_name"], route_compliance["avg(compliant)"])
plt.xlabel("Compliance rate")
plt.ylabel("Route")
plt.title("Figure 1: Compliance Rate by Route")
plt.tight_layout()
plt.savefig("/home/sonu/BIG_DATA/outputs/fig1_compliance_by_route.png")
plt.show()
```

```python
deviation_sample = df.select("deviation_minutes").sample(0.1, seed=42).toPandas()

plt.figure(figsize=(8, 5))
plt.hist(deviation_sample["deviation_minutes"], bins=40)
plt.xlabel("Deviation from schedule (minutes)")
plt.ylabel("Count")
plt.title("Figure 2: Distribution of Schedule Deviation")
plt.tight_layout()
plt.savefig("/home/sonu/BIG_DATA/outputs/fig2_deviation_distribution.png")
plt.show()
```

```python
geo_sample = df.select("latitude", "longitude", "compliant").sample(0.2, seed=42).toPandas()

plt.figure(figsize=(8, 8))
compliant_pts = geo_sample[geo_sample["compliant"] == 1]
noncompliant_pts = geo_sample[geo_sample["compliant"] == 0]

plt.scatter(compliant_pts["longitude"], compliant_pts["latitude"], s=5, alpha=0.4, label="compliant")
plt.scatter(noncompliant_pts["longitude"], noncompliant_pts["latitude"], s=5, alpha=0.4, label="non-compliant")
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.title("Figure 3: Compliant vs Non-Compliant Trips - Medway/Kent Network")
plt.legend()
plt.tight_layout()
plt.savefig("/home/sonu/BIG_DATA/outputs/fig3_compliance_map.png")
plt.show()
```

```python
compliance_totals = df.groupBy("compliant").count().toPandas()
compliance_totals["label"] = compliance_totals["compliant"].map({1: "Compliant", 0: "Non-Compliant"})

plt.figure(figsize=(6, 6))
plt.pie(compliance_totals["count"], labels=compliance_totals["label"], autopct="%1.1f%%",
        colors=["#55A868", "#C44E52"], startangle=90)
plt.title("Figure 4: Overall Compliance Split")
plt.tight_layout()
plt.savefig("/home/sonu/BIG_DATA/outputs/fig4_overall_compliance_pie.png")
plt.show()
```

```python
from pyspark.sql.functions import hour, col

hourly_compliance = df.withColumn("hour_of_day", hour(col("scheduled_datetime"))) \
    .groupBy("hour_of_day").avg("compliant").orderBy("hour_of_day").toPandas()

plt.figure(figsize=(9, 5))
plt.plot(hourly_compliance["hour_of_day"], hourly_compliance["avg(compliant)"], marker="o", color="#DD8452")
plt.xlabel("Hour of Day")
plt.ylabel("Compliance Rate")
plt.title("Figure 5: Compliance Rate by Hour of Day")
plt.tight_layout()
plt.savefig("/home/sonu/BIG_DATA/outputs/fig5_hourly_compliance_line.png")
plt.show()
```
